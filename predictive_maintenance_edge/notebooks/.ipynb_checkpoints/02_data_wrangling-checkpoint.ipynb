```python
# 3. Features and target
X = df1[
    ["revolutions", "humidity"]
]

y = df1["vibration"]
```

```python
# 4. Chronological split TRAIN / VALIDATION / TEST 
#    60% Train
#    20% Validation
#    20% Test
# =========================================================

n = len(df1)

train_end = int(n * 0.60)
val_end = int(n * 0.80)



X_train = X.iloc[:train_end]
y_train = y.iloc[:train_end]

X_val = X.iloc[train_end:val_end]
y_val = y.iloc[train_end:val_end]


X_test = X.iloc[val_end:]
y_test = y.iloc[val_end:]

print("\nDataset sizes")

print(X_train)
print("=====================================================")
print(X_test)
print("=====================================================")
print(y_train)
print("=====================================================")
print(y_test)

print("Training rows:", len(X_train))
print("Validation rows:", len(X_val))
print("Test rows:", len(X_test))
```

```python
# 5. Train model
model = LinearRegression()

model.fit(
    X_train,
    y_train
)
```

```python
# 6. Predictions
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)
print(train_pred)
print("=================================")
print(test_pred)
```

```python
# 11. Severity classification
def vibration_status(residual):
    error = abs(residual)

    if error <= threshold:
        return "Normal"

    elif error <= 2 * threshold:
        return "Warning"

    else:
        return "Critical"

results["status"] = (
    results["residual"]
    .apply(vibration_status)
)
```
